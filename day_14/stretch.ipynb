```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(42)

# ============================================================
# 1. Poisson → Exponential Link
# ============================================================

gaps = rng.exponential(scale=1, size=1000)

plt.figure(figsize=(7, 4))
plt.hist(gaps, bins=30, density=True, alpha=0.6, label="Simulated gaps")

x = np.linspace(0, 8, 300)
plt.plot(x, stats.expon.pdf(x), label="Exponential PDF")

plt.title("Poisson Arrivals → Exponential Gaps")
plt.xlabel("Time gap")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()

print("1. Poisson arrivals have exponentially distributed waiting times.")


# ============================================================
# 2. Binomial → Poisson
# ============================================================

n = 1000
p = 0.01
lam = n * p

x = np.arange(0, 20)

binomial = stats.binom.pmf(x, n, p)
poisson = stats.poisson.pmf(x, lam)

plt.figure(figsize=(7, 4))
plt.plot(x, binomial, "o-", label="Binomial")
plt.plot(x, poisson, "s--", label="Poisson")

plt.title("Binomial vs Poisson")
plt.xlabel("Number of events") 
plt.ylabel("Probability")
plt.legend()
plt.grid(True)
plt.show()

print("2. For large n and small p, Binomial approaches Poisson.")


# ============================================================
# 3. Inverse-Transform Sampling
# ============================================================

u = rng.random(1000)

# Inverse CDF of Exponential distribution
samples = -np.log(1 - u)

plt.figure(figsize=(7, 4))
plt.hist(samples, bins=30, density=True, alpha=0.6)
plt.title("Inverse-Transform Sampling")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.show()

print("3. Uniform random numbers can generate other distributions using inverse CDF.")


# ============================================================
# 4. Realistic Income Dataset
# ============================================================

income = rng.lognormal(mean=10, sigma=0.6, size=1000)

plt.figure(figsize=(7, 4))
plt.hist(income, bins=30)
plt.title("Realistic Income Distribution")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()

print("4. Income is often right-skewed because a smaller group has much higher incomes.")


# ============================================================
# 5. Log-Normal Distribution
# ============================================================

lognormal_data = rng.lognormal(mean=2, sigma=0.8, size=1000)

plt.figure(figsize=(7, 4))
plt.hist(lognormal_data, bins=30, density=True)
plt.title("Log-Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.grid(True)
plt.show()

log_data = np.log(lognormal_data)

plt.figure(figsize=(7, 4))
plt.hist(log_data, bins=30, density=True)
plt.title("Log of Log-Normal Data")
plt.xlabel("log(Value)")
plt.ylabel("Density")
plt.grid(True)
plt.show()

print("5. A log-normal distribution becomes approximately normal after taking log.")
```
