```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# ============================================================
# STRETCH TASKS — CALCULUS
# ============================================================

x, y, z = sp.symbols("x y z")


# ============================================================
# 1. Numerical derivative vs SymPy
#    Try several values of h and plot error
# ============================================================

print("=" * 60)
print("1. NUMERICAL DERIVATIVE ERROR")
print("=" * 60)

f = lambda x: x**4
exact_derivative = lambda x: 4*x**3

def numerical_derivative(func, x, h):
    return (func(x + h) - func(x - h)) / (2 * h)

x_value = 2

# Different step sizes
h_values = np.logspace(-1, -15, 15)

errors = []

for h in h_values:
    numerical = numerical_derivative(f, x_value, h)
    exact = exact_derivative(x_value)
    error = abs(numerical - exact)
    errors.append(error)

print("x =", x_value)
print("Exact derivative =", exact_derivative(x_value))

for h, error in zip(h_values, errors):
    print(f"h = {h:.1e}, error = {error:.6e}")

# Plot error
plt.figure(figsize=(8, 5))
plt.loglog(h_values, errors, marker="o")
plt.xlabel("Step size h")
plt.ylabel("Absolute error")
plt.title("Numerical Derivative Error vs Step Size")
plt.grid(True)
plt.show()

print(
    "\nWhy does the error increase when h is too small?"
)
print(
    "When h becomes extremely small, floating-point rounding "
    "and subtraction errors become significant, so the numerical "
    "derivative becomes less accurate."
)


# ============================================================
# 2. Three-variable function and gradient
# ============================================================

print("\n" + "=" * 60)
print("2. THREE-VARIABLE PARTIAL DERIVATIVES")
print("=" * 60)

f3 = x**2 + 2*y**2 + 3*z**2 + x*y + y*z + x*z

# By hand:
#
# ∂f/∂x = 2x + y + z
# ∂f/∂y = 4y + x + z
# ∂f/∂z = 6z + y + x

df_dx = sp.diff(f3, x)
df_dy = sp.diff(f3, y)
df_dz = sp.diff(f3, z)

print("f(x,y,z) =", f3)

print("\nBy hand:")
print("∂f/∂x = 2x + y + z")
print("∂f/∂y = 4y + x + z")
print("∂f/∂z = 6z + y + x")

print("\nSymPy:")
print("∂f/∂x =", df_dx)
print("∂f/∂y =", df_dy)
print("∂f/∂z =", df_dz)


def gradient_3d(px, py, pz):
    return np.array([
        2*px + py + pz,
        4*py + px + pz,
        6*pz + py + px
    ])


grad = gradient_3d(1, 2, 3)

print("\nGradient at (1, 2, 3):")
print(grad)

print("\nGradient type:", type(grad))
print("Gradient shape:", grad.shape)
print("Number of components:", len(grad))

print("\nThe gradient is a 3-vector because the function has three variables.")


# ============================================================
# 3. Visualise a saddle
#    f(x,y) = x² - y²
# ============================================================

print("\n" + "=" * 60)
print("3. SADDLE FUNCTION")
print("=" * 60)

X, Y = np.meshgrid(
    np.linspace(-3, 3, 20),
    np.linspace(-3, 3, 20)
)

Z = X**2 - Y**2

# Gradient:
# ∂f/∂x = 2x
# ∂f/∂y = -2y

U = 2*X
V = -2*Y

# Normalize arrows
magnitude = np.sqrt(U**2 + V**2)
U_norm = U / (magnitude + 1e-10)
V_norm = V / (magnitude + 1e-10)

plt.figure(figsize=(9, 7))

contours = plt.contour(X, Y, Z, levels=20)
plt.clabel(contours, fontsize=8)

plt.quiver(
    X,
    Y,
    U_norm,
    V_norm,
    angles="xy",
    scale=25
)

plt.scatter(
    0,
    0,
    s=100,
    marker="*",
    label="Saddle point (0,0)"
)

plt.axhline(0)
plt.axvline(0)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Saddle: f(x,y) = x² - y²")
plt.grid(True)
plt.legend()
plt.show()

print("Near the centre, the gradient arrows point:")
print("- Along the x-direction, away from x = 0.")
print("- Along the y-direction, toward y = 0.")
print(
    "The centre is neither a peak nor a valley because the function "
    "increases in the x-direction but decreases in the y-direction."
)


# ============================================================
# 4. Four-layer chain rule
#    f(x) = cos(sin(x² + 1))
# ============================================================

print("\n" + "=" * 60)
print("4. FOUR-LAYER CHAIN RULE")
print("=" * 60)

f4 = sp.cos(sp.sin(x**2 + 1))

df4 = sp.diff(f4, x)

print("Function:")
print("f(x) = cos(sin(x² + 1))")

print("\nFour layers:")
print("1. x")
print("2. x² + 1")
print("3. sin(x² + 1)")
print("4. cos(sin(x² + 1))")

print("\nBy hand:")
print("f'(x) = -sin(sin(x² + 1)) * cos(x² + 1) * 2x")

print("\nSymPy:")
print("f'(x) =", df4)

# Verify individual stages with SymPy
u = x**2 + 1
v = sp.sin(u)
w = sp.cos(v)

print("\n=== Individual stages ===")
print("du/dx =", sp.diff(u, x))
print("dv/dx =", sp.diff(v, x))
print("dw/dx =", sp.diff(w, x))

print(
    "\nThe final derivative matches the chain rule applied "
    "from the outside toward the inside."
)


# ============================================================
# 5. Gradient is the steepest direction
# ============================================================

print("\n" + "=" * 60)
print("5. GRADIENT = STEEPEST DIRECTION")
print("=" * 60)

def function_2d(px, py):
    return px**2 + 2*py**2 + px*py


def gradient(px, py):
    return np.array([
        2*px + py,
        4*py + px
    ])


# Choose a point
px = 1
py = 1

grad = gradient(px, py)
grad_magnitude = np.linalg.norm(grad)

print("Point:", (px, py))
print("Gradient:", grad)
print("Gradient magnitude:", grad_magnitude)

# Normalize gradient to compare directions fairly
gradient_direction = grad / np.linalg.norm(grad)

# Different unit directions
directions = {
    "Gradient direction": gradient_direction,
    "x direction": np.array([1.0, 0.0]),
    "y direction": np.array([0.0, 1.0]),
    "Diagonal direction": np.array([1.0, 1.0]) / np.sqrt(2),
    "Opposite gradient": -gradient_direction
}

step = 0.001

original_value = function_2d(px, py)

print("\nFunction value at original point:", original_value)
print("\nIncrease for different directions:")

for name, direction in directions.items():

    new_x = px + step * direction[0]
    new_y = py + step * direction[1]

    new_value = function_2d(new_x, new_y)

    increase = new_value - original_value

    print(f"{name:22s}: increase = {increase:.8f}")

print("\nConclusion:")
print(
    "For unit-length directions, the gradient direction gives the "
    "largest increase in the function."
)

print(
    "Therefore, the gradient points in the direction of the "
    "steepest ascent."
)
```
