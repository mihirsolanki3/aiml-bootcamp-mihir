Part A — derivatives

```python
# TASK 01
# Compute derivatives by hand, verify with sympy

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Symbol
x = sp.symbols("x")

# -------------------------------------------------
# TASK 01: Derivatives
# -------------------------------------------------

# Functions
n = 4
f = x**n
g = 3*x**2 + 2*x - 7
h = 1/x

# Derivatives using SymPy
df = sp.diff(f, x)
dg = sp.diff(g, x)
dh = sp.diff(h, x)

print("=== Exact Derivatives ===")
print("f(x) =", f)
print("f'(x) =", df)

print("\ng(x) =", g)
print("g'(x) =", dg)

print("\nh(x) =", h)
print("h'(x) =", dh)

# Evaluate derivatives at x = 2
x_value = 2

print("\n=== Derivatives at x = 2 ===")
print("f'(2) =", df.subs(x, x_value))
print("g'(2) =", dg.subs(x, x_value))
print("h'(2) =", dh.subs(x, x_value))

# -------------------------------------------------
# Numerical derivative
# -------------------------------------------------

def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x - h)) / (2 * h)

# Numerical functions
f_num = lambda x: x**4
g_num = lambda x: 3*x**2 + 2*x - 7
h_num = lambda x: 1/x

# Numerical derivatives at x = 2
f_numerical = numerical_derivative(f_num, 2)
g_numerical = numerical_derivative(g_num, 2)
h_numerical = numerical_derivative(h_num, 2)

print("\n=== Numerical Derivatives at x = 2 ===")
print("f'(2) =", f_numerical)
print("g'(2) =", g_numerical)
print("h'(2) =", h_numerical)

# -------------------------------------------------
# Compare exact and numerical values
# -------------------------------------------------

print("\n=== Comparison ===")
print("f'(2): Exact =", float(df.subs(x, 2)),
      "| Numerical =", f_numerical)

print("g'(2): Exact =", float(dg.subs(x, 2)),
      "| Numerical =", g_numerical)

print("h'(2): Exact =", float(dh.subs(x, 2)),
      "| Numerical =", h_numerical)

# -------------------------------------------------
# Find where derivative of x^n equals zero
# -------------------------------------------------

zero_points = sp.solve(df, x)

print("\n=== Where f'(x) = 0 ===")
print("f'(x) =", df)
print("Derivative equals zero at x =", zero_points)
print("At this point, the function has a horizontal tangent.")

# -------------------------------------------------
# Plot f(x) = x^4
# -------------------------------------------------

x_values = np.linspace(-2, 2, 400)
y_values = x_values**4

plt.plot(x_values, y_values)
plt.axhline(0)
plt.axvline(0)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("f(x) = x⁴")
plt.grid(True)
plt.show()
```

```python
# TASK 02
# Plot a function against its derivative

import numpy as np
import matplotlib.pyplot as plt

# TASK 02: Plot a function against its derivative

# x values
x = np.linspace(-3, 3, 200)

# Function: f(x) = x^3 - 3x
f = x**3 - 3*x

# Derivative: f'(x) = 3x^2 - 3
df = 3*x**2 - 3

# Points where derivative = 0
# 3x^2 - 3 = 0
# x = -1, 1
zero_points = np.array([-1, 1])

# Function values at the zero points
f_zero = zero_points**3 - 3*zero_points

# Create two panels sharing x-axis
fig, axes = plt.subplots(2, 1, figsize=(8, 8), sharex=True)

# -------------------------------------------------
# Top plot: Original function
# -------------------------------------------------

axes[0].plot(x, f, label="f(x) = x³ - 3x")
axes[0].scatter(
    zero_points,
    f_zero,
    s=80,
    zorder=5,
    label="Peak / Valley"
)

axes[0].axhline(0)
axes[0].axvline(0)
axes[0].set_ylabel("f(x)")
axes[0].set_title("Function and Its Derivative")
axes[0].grid(True)
axes[0].legend()

# -------------------------------------------------
# Bottom plot: Derivative
# -------------------------------------------------

axes[1].plot(x, df, label="f'(x) = 3x² - 3")
axes[1].scatter(
    zero_points,
    [0, 0],
    s=80,
    zorder=5,
    label="Derivative = 0"
)

axes[1].axhline(0)
axes[1].axvline(0)
axes[1].set_xlabel("x")
axes[1].set_ylabel("f'(x)")
axes[1].grid(True)
axes[1].legend()

# Mark zero points
for point in zero_points:
    axes[1].annotate(
        f"x = {point}",
        (point, 0),
        xytext=(10, 15),
        textcoords="offset points"
    )

plt.tight_layout()
plt.show()

# -------------------------------------------------
# Confirm peak and valley
# -------------------------------------------------

print("=== Critical Points ===")

for point, value in zip(zero_points, f_zero):
    print(f"x = {point}, f(x) = {value}")

print("\nAt x = -1: local maximum (peak)")
print("At x =  1: local minimum (valley)")

print("\n=== Conclusion ===")
print(
    "From the derivative plot, peaks and valleys occur where "
    "the derivative crosses zero and changes sign."
)
```

Part B — gradients and the chain rule

```python
# TASK 03
# Partial derivatives and the gradient

import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# ============================================================
# TASK 03: Partial Derivatives and Gradient
# ============================================================

# ------------------------------------------------------------
# 1. Define the function using SymPy
# ------------------------------------------------------------

x, y = sp.symbols("x y")

f = x**2 + 2*y**2 + x*y

# By hand:
# ∂f/∂x = 2x + y
# ∂f/∂y = 4y + x

# Verify using SymPy
df_dx = sp.diff(f, x)
df_dy = sp.diff(f, y)

print("=== Function ===")
print("f(x, y) =", f)

print("\n=== Partial Derivatives ===")
print("By hand:")
print("∂f/∂x = 2x + y")
print("∂f/∂y = 4y + x")

print("\nSymPy:")
print("∂f/∂x =", df_dx)
print("∂f/∂y =", df_dy)


# ------------------------------------------------------------
# 2. Gradient function
# ------------------------------------------------------------

def gradient(px, py):
    return np.array([
        2 * px + py,
        4 * py + px
    ])


# Function for numerical evaluation
def function(px, py):
    return px**2 + 2*py**2 + px*py


# ------------------------------------------------------------
# 3. Evaluate gradient at three points
# ------------------------------------------------------------

points = [
    (1, 1),
    (-1, 1),
    (2, -1)
]

step = 0.01

print("\n=== Gradient Checks ===")

for px, py in points:

    grad = gradient(px, py)

    # Original function value
    original = function(px, py)

    # Step in gradient direction
    uphill = function(
        px + step * grad[0],
        py + step * grad[1]
    )

    # Step in opposite direction
    downhill = function(
        px - step * grad[0],
        py - step * grad[1]
    )

    print(f"\nPoint ({px}, {py})")
    print("Gradient =", grad)
    print("Original f =", original)
    print("Along gradient =", uphill)
    print("Against gradient =", downhill)

    print("Uphill check:", uphill > original)
    print("Downhill check:", downhill < original)


# ------------------------------------------------------------
# 4. Contour plot + gradient arrows
# ------------------------------------------------------------

# Create grid
X, Y = np.meshgrid(
    np.linspace(-3, 3, 20),
    np.linspace(-3, 3, 20)
)

# Function values
Z = X**2 + 2*Y**2 + X*Y

# Gradient components
U = 2*X + Y
V = 4*Y + X

# Normalize arrows so they are easier to see
magnitude = np.sqrt(U**2 + V**2)

U_normalized = U / (magnitude + 1e-10)
V_normalized = V / (magnitude + 1e-10)

# Plot
plt.figure(figsize=(9, 7))

contours = plt.contour(
    X,
    Y,
    Z,
    levels=15
)

plt.clabel(contours, inline=True, fontsize=8)

plt.quiver(
    X,
    Y,
    U_normalized,
    V_normalized,
    angles="xy",
    scale=25
)

# Minimum
plt.scatter(
    0,
    0,
    s=100,
    marker="*",
    label="Minimum (0, 0)"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Contour Plot with Gradient Vectors")
plt.axhline(0)
plt.axvline(0)
plt.grid(True)
plt.legend()
plt.show()


# ------------------------------------------------------------
# 5. Final conclusion
# ------------------------------------------------------------

print("\n=== Conclusion ===")
print(
    "The gradient points in the direction of greatest increase "
    "of the function, so the gradient arrows point away from "
    "the minimum at (0, 0)."
)
```

```python
# TASK 04
# The chain rule, by hand and in code

import numpy as np
import sympy as sp

# ============================================================
# TASK 04: The Chain Rule, by Hand and in Code
# ============================================================

x = sp.symbols("x")

# ------------------------------------------------------------
# PART 1: f(x) = (3x + 1)^5
# ------------------------------------------------------------

# Outer function:
# u^5
#
# Inner function:
# u = 3x + 1
#
# Chain rule:
# df/dx = 5(3x + 1)^4 * 3
#       = 15(3x + 1)^4

print("=== PART 1: (3x + 1)^5 ===")

f1 = (3*x + 1)**5

# SymPy derivative
df1 = sp.diff(f1, x)

print("Function:")
print("f(x) =", f1)

print("\nBy hand:")
print("Outer function: u^5")
print("Inner function: u = 3x + 1")
print("f'(x) = 5(3x + 1)^4 * 3")
print("f'(x) = 15(3x + 1)^4")

print("\nSymPy:")
print("f'(x) =", df1)


# ------------------------------------------------------------
# Tiny-step numerical derivative
# ------------------------------------------------------------

def numerical_derivative(func, x, h=1e-5):
    return (func(x + h) - func(x - h)) / (2 * h)


f1_num = lambda x: (3*x + 1)**5

test_values = [0, 1]

print("\n=== Numerical Verification ===")

for value in test_values:

    exact = float(df1.subs(x, value))
    numerical = numerical_derivative(f1_num, value)

    print(f"\nx = {value}")
    print("Exact derivative     =", exact)
    print("Numerical derivative =", numerical)
    print("Difference           =", abs(exact - numerical))


# ------------------------------------------------------------
# PART 2: Three-layer chain
# f(x) = sin((2x + 1)^2)
# ------------------------------------------------------------

print("\n\n=== PART 2: sin((2x + 1)^2) ===")

f2 = sp.sin((2*x + 1)**2)

df2 = sp.diff(f2, x)

print("Function:")
print("f(x) = sin((2x + 1)^2)")

print("\nBy hand using chain rule:")
print("Outer: sin(u)")
print("Middle: u = v^2")
print("Inner: v = 2x + 1")

print("\nf'(x) = cos((2x + 1)^2) * 2(2x + 1) * 2")
print("f'(x) = 4(2x + 1)cos((2x + 1)^2)")

print("\nSymPy:")
print("f'(x) =", df2)

# ------------------------------------------------------------
# Numerical verification of three-layer chain
# ------------------------------------------------------------

f2_num = lambda x: np.sin((2*x + 1)**2)

test_values_2 = [0, 1]

print("\n=== Three-Layer Numerical Verification ===")

for value in test_values_2:

    exact = float(df2.subs(x, value))
    numerical = numerical_derivative(f2_num, value)

    print(f"\nx = {value}")
    print("Exact derivative     =", exact)
    print("Numerical derivative =", numerical)
    print("Difference           =", abs(exact - numerical))

# ------------------------------------------------------------
# Final explanation
# ------------------------------------------------------------

print("\n=== Explanation ===")

print(
    "The chain rule multiplies derivatives because a change in the "
    "input causes a change in the inner function, which then causes "
    "a change in the outer function."
)

print(
    "Multiplying the rates of change connects these changes together, "
    "giving the total rate of change from the original input to the "
    "final output."
)
```
