```python
#  11. Compute the sum of each row, and separately the sum of each column.

# axis=0 works DOWN the rows, giving the sum of each column.
# axis=1 works ACROSS the columns, giving the sum of each row.

row_sums = grid.sum(axis=1)
column_sums = grid.sum(axis=0)

print("Row sums:", row_sums)
print("Column sums:", column_sums)
```

```python
#  12. Normalise a 2D array so every column has mean 0 and standard deviation 1.

data = np.array([
    [10, 100, 1000],
    [20, 200, 2000],
    [30, 300, 3000],
    [40, 400, 4000]
], dtype=float)

column_mean = data.mean(axis=0)
column_std = data.std(axis=0)

normalized = (data - column_mean) / column_std

print("Original:")
print(data)

print("\nNormalized:")
print(normalized)

print("\nColumn means:")
print(normalized.mean(axis=0))

print("\nColumn standard deviations:")
print(normalized.std(axis=0))
```

TASK 03
Convert loops to NumPy and measure the speedup

```python
import numpy as np
import timeit
import matplotlib.pyplot as plt

# Array sizes
sizes = [1_000, 10_000, 100_000, 1_000_000]

results = []

# Python loop operations
def square_python(lst):
    return [x * x for x in lst]

def sum_python(lst):
    total = 0
    for x in lst:
        total += x
    return total

def dot_python(a, b):
    total = 0
    for x, y in zip(a, b):
        total += x * y
    return total


# NumPy operations
def square_numpy(arr):
    return arr ** 2

def sum_numpy(arr):
    return np.sum(arr)

def dot_numpy(a, b):
    return np.dot(a, b)


# Measure performance
for n in sizes:
    lst = list(range(n))
    lst2 = list(range(n))
    
    arr = np.arange(n)
    arr2 = np.arange(n)

    number = 10

    # Squaring
    py_square = timeit.timeit(
        lambda: square_python(lst), number=number
    ) / number

    np_square = timeit.timeit(
        lambda: square_numpy(arr), number=number
    ) / number

    # Sum
    py_sum = timeit.timeit(
        lambda: sum_python(lst), number=number
    ) / number

    np_sum = timeit.timeit(
        lambda: sum_numpy(arr), number=number
    ) / number

    # Dot product
    py_dot = timeit.timeit(
        lambda: dot_python(lst, lst2), number=number
    ) / number

    np_dot = timeit.timeit(
        lambda: dot_numpy(arr, arr2), number=number
    ) / number

    results.append({
        "Size": n,
        "Square Speedup": py_square / np_square,
        "Sum Speedup": py_sum / np_sum,
        "Dot Speedup": py_dot / np_dot
    })


# Display results table
print("=" * 80)
print("NUMPY vs PYTHON LOOP SPEEDUP")
print("=" * 80)

print(f"{'Size':>12} {'Square':>15} {'Sum':>15} {'Dot Product':>15}")
print("-" * 80)

for r in results:
    print(
        f"{r['Size']:>12,} "
        f"{r['Square Speedup']:>14.2f}x "
        f"{r['Sum Speedup']:>14.2f}x "
        f"{r['Dot Speedup']:>14.2f}x"
    )

# Plot speedup
x = [r["Size"] for r in results]
square_speedup = [r["Square Speedup"] for r in results]
sum_speedup = [r["Sum Speedup"] for r in results]
dot_speedup = [r["Dot Speedup"] for r in results]

plt.figure(figsize=(10, 6))

plt.plot(x, square_speedup, marker="o", label="Squaring")
plt.plot(x, sum_speedup, marker="o", label="Sum")
plt.plot(x, dot_speedup, marker="o", label="Dot Product")

plt.xscale("log")
plt.xlabel("Array Size")
plt.ylabel("Speedup (Python / NumPy)")
plt.title("NumPy Speedup vs Python Loops")
plt.grid(True)
plt.legend()

plt.show()
```

TASK 04
Broadcasting with mismatched shapes

```python
# 1. Create a (3, 4) array and a (4,) array. Add them. Explain in a comment why it works.

a = np.ones((3, 4))
b = np.array([1, 2, 3, 4])

result = a + b

print("a shape:", a.shape)
print("b shape:", b.shape)
print("Result:\n", result)

# It works because NumPy broadcasts the (4,) array across
# all 3 rows of the (3, 4) array.
```

```python
# 2. Create a (3, 4) array and a (3,) array. Add them. It fails — read the error and write down what shapes NumPycompared.

a = np.ones((3, 4))
c = np.array([1, 2, 3])

try:
    result = a + c
except ValueError as e:
    print("\nBroadcasting error:")
    print(e)

# (3, 4) and (3,) do not match.
# NumPy compares dimensions from RIGHT to LEFT:
# (3, 4)
#    (3)
# The dimensions 4 and 3 are different, so broadcasting fails.
```

```python
# 3. Fix that failure with reshape. Explain what the reshape changed

c_fixed = c.reshape(3, 1)

result = a + c_fixed

print("\nFixed shape:", c_fixed.shape)
print("Result:\n", result)

# reshape changes c from (3,) to (3, 1).
# Now NumPy compares:
# (3, 4)
# (3, 1)
# The 1 can be broadcast to 4, so the addition works.
```

```python
# 4. Add a (3, 1) array to a (1, 4) array. Predict the output shape before running it. Were you right?

x = np.array([[1],
              [2],
              [3]])

y = np.array([[10, 20, 30, 40]])

result = x + y

print("\n(3, 1) + (1, 4)")
print("Predicted output shape: (3, 4)")
print("Actual output shape:", result.shape)
print(result)

# Prediction: (3, 4)
# Yes, the prediction was correct.
```

TASK 06
A real analysis, no loops

```python
# 1. Generate a dataset: 100 students × 5 exams of random marks between 30 and 100. Seed your generator so results are reproducible.

rng = np.random.default_rng(42)

marks = rng.integers(30, 101, size=(100, 5))

print("Dataset shape:", marks.shape)
print("Marks:\n", marks)
```

```python
# 2. Compute each student's average and each exam's average. State which axis you used for each.

# axis=1 → calculate across columns (5 exams)
# Therefore, we get one average for each student
per_student = marks.mean(axis=1)

# axis=0 → calculate down rows (100 students)
# Therefore, we get one average for each exam
per_exam = marks.mean(axis=0)

print("\nAverage of each student:")
print(per_student)

print("\nAverage of each exam:")
print(per_exam)
```

```python
# 3. Find how many students passed (average ‡ 60), and what fraction that is.

# A student passes if their average >= 60
passed_mask = per_student >= 60

passed_count = np.sum(passed_mask)
passed_fraction = np.mean(passed_mask)

print("\nStudents passed:", passed_count)
print("Fraction passed:", passed_fraction)
```

```python
# 4. Find the top 10 students by average — return their indices and their scores.

# argsort returns indices that would sort the array
# [::-1] reverses it → highest average first
top_10_indices = np.argsort(per_student)[::-1][:10]

top_10_scores = per_student[top_10_indices]

print("\nTop 10 students:")
print("Student indices:", top_10_indices)
print("Student averages:", top_10_scores)
```

```python
#  5. Standardise the marks per exam (mean 0, std 1), so exams of different difficulty become comparable.

# Calculate mean and standard deviation for each exam
exam_mean = marks.mean(axis=0)
exam_std = marks.std(axis=0)

# Broadcasting applies each exam's mean/std to its column
standardised = (marks - exam_mean) / exam_std

print("\nStandardised marks:")
print(standardised)

# Check: each exam should have mean ≈ 0 and std ≈ 1
print("\nStandardised exam means:")
print(standardised.mean(axis=0))

print("\nStandardised exam standard deviations:")
print(standardised.std(axis=0))

```

```python
# 6. Apply a curve: add 5 marks to everyone but cap at 100. Report how many marks hit the cap

# Add 5 marks to everyone
# np.minimum caps every mark at 100
curved_marks = np.minimum(marks + 5, 100)

# Count how many marks reached the cap
cap_count = np.sum(curved_marks == 100)

print("\nCurved marks:")
print(curved_marks)

print("\nNumber of marks that hit 100:", cap_count)
```

```python
# 7. Identify any student who failed every single exam. (Hint: a mask, then an all-along-an-axis check.)

# Failed exam = mark < 60
failed_exam = marks < 60

# axis=1 → check all 5 exams for each student
# True only when ALL exams are below 60
failed_every_exam = np.all(failed_exam, axis=1)

failed_students = np.where(failed_every_exam)[0]

print("\nStudents who failed every exam:")
print(failed_students)

print("Number of students who failed every exam:",
      len(failed_students))
```
