```python
import numpy as np
import seaborn as sns

# Reproducible data
rng = np.random.default_rng(42)

# ---------------------------------------------------------
# 1. Recreate an analysis result as a polished figure
# ---------------------------------------------------------

students = np.arange(1, 11)
study_hours = np.array([2, 3, 4, 5, 6, 7, 8, 4, 6, 9])
marks = np.array([45, 50, 58, 65, 70, 78, 85, 62, 74, 92])

fig, ax = plt.subplots(figsize=(9, 6))

ax.scatter(study_hours, marks, s=70, alpha=0.8)

ax.set_title("Relationship Between Study Hours and Marks",
             fontsize=16, fontweight="bold")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Marks")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()

```

```python

# ---------------------------------------------------------
# 2. Explore plt.style.use()
# ---------------------------------------------------------

# Try these one at a time:
# plt.style.use("seaborn-v0_8")
# plt.style.use("ggplot")

import matplotlib.pyplot as plt

plt.style.use("dark_background")

plt.figure(figsize=(8, 5))
plt.plot(study_hours, marks, marker="o") 
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Technical report:
# "seaborn-v0_8" is generally the best choice for a technical report
# because it gives clean, readable plots without being too distracting.


```

```python
# ---------------------------------------------------------
# 3. Add a trend line using np.polyfit
# ---------------------------------------------------------

# Calculate linear trend line
slope, intercept = np.polyfit(study_hours, marks, 1)

trend_line = slope * study_hours + intercept

plt.figure(figsize=(9, 6))

plt.scatter(
    study_hours,
    marks,
    s=70,
    alpha=0.8,
    label="Observed data"
)

plt.plot(
    study_hours,
    trend_line,
    linewidth=2,
    label="Trend line"
)

plt.title("Study Hours vs Marks with Trend Line",
          fontsize=16, fontweight="bold")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Slope: {slope:.2f}")
print(f"Intercept: {intercept:.2f}")


```

```python
# ---------------------------------------------------------
# 4. Annotated chart
# ---------------------------------------------------------

highest_index = np.argmax(marks)

plt.figure(figsize=(9, 6))

plt.scatter(study_hours, marks, s=70, alpha=0.8)

plt.annotate(
    "Highest mark",
    xy=(study_hours[highest_index], marks[highest_index]),
    xytext=(study_hours[highest_index] - 2, marks[highest_index] - 10),
    arrowprops=dict(arrowstyle="->", linewidth=2),
    fontsize=12
)

plt.title("Annotated Study Performance",
          fontsize=16, fontweight="bold")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()


```

```python
# ---------------------------------------------------------
# 5. Histogram using Matplotlib
# ---------------------------------------------------------

marks_data = rng.normal(70, 12, 100)

plt.figure(figsize=(9, 6))

plt.hist(
    marks_data,
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Student Marks",
          fontsize=16, fontweight="bold")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()


```

```python
# ---------------------------------------------------------
# 6. Same histogram using Seaborn
# ---------------------------------------------------------

plt.figure(figsize=(9, 6))

sns.histplot(
    marks_data,
    bins=10,
    kde=True
)

plt.title("Distribution of Student Marks — Seaborn",
          fontsize=16, fontweight="bold")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# 7. Conclusion
# ---------------------------------------------------------

print("""
Stretch Task Conclusions
-------------------------
1. A polished figure communicates numerical analysis more clearly.
2. plt.style.use() changes the overall appearance of Matplotlib plots.
3. seaborn-v0_8 is a good choice for technical reports because it is clean
   and readable.
4. np.polyfit() can be used to calculate a linear trend line.
5. ax.annotate() helps highlight important observations.
6. Seaborn requires less code for many statistical plots.
7. Seaborn is useful when creating statistical visualisations quickly,
   while Matplotlib provides more detailed control over individual elements.
""")
```
